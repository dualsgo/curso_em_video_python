# Modularização e Pacotes em Python

A **modularização** surgiu no início da década de 1960 para lidar com programas cada vez maiores. O foco principal é **dividir um programa grande em partes menores**.

### Vantagens da Modularização:
1. **Organização do código**: separar funções e classes por responsabilidade.
2. **Facilidade na manutenção**: corrigir um bug em um módulo atualiza todos os programas que o utilizam.
3. **Ocultação de código detalhado (Abstração)**: quem usa a função não precisa ver a complexidade interna dela.
4. **Reutilização em outros projetos**: basta copiar o módulo ou pacote para outro projeto.

## 1. Criando o seu próprio Módulo

Em Python, **qualquer arquivo `.py` é um módulo**!

Por exemplo, se criarmos um arquivo chamado `moeda.py` na mesma pasta do nosso programa principal (`programa.py`):

```python
# Arquivo: moeda.py
def aumentar(preco=0, taxa=0):
    return preco + (preco * taxa / 100)

def diminuir(preco=0, taxa=0):
    return preco - (preco * taxa / 100)

def dobro(preco=0):
    return preco * 2

def metade(preco=0):
    return preco / 2
```

No arquivo principal (`programa.py`), podemos importar esse módulo de duas maneiras:
- `import moeda` → Acessamos com `moeda.dobro(100)` (**Mais recomendado** para evitar conflito de nomes).
- `from moeda import dobro, metade` → Acessamos diretamente com `dobro(100)`.

## 2. A variável especial `__name__` e o bloco `if __name__ == '__main__':`

Quando você executa um arquivo `.py` diretamente, o Python define internamente a variável especial `__name__` daquele arquivo com o valor `'__main__'`.

Porém, quando esse mesmo arquivo é **importado como módulo** dentro de outro arquivo, o `__name__` dele passa a ser o **próprio nome do arquivo** (ex: `'moeda'`).

Por isso, em projetos profissionais e em POO, colocamos os testes ou a execução principal dentro de `if __name__ == '__main__':`. Assim, o código de teste só roda se executarmos o arquivo diretamente, e **não** dispara por acidente quando importamos suas funções ou classes em outro lugar!

````python
def somar(a: int, b: int) -> int:
    return a + b

print('Valor de __name__ neste ambiente:', __name__)

# Tudo dentro deste if só executa quando o arquivo é rodado diretamente (não quando é importado):
if __name__ == '__main__':
    print('Executando teste local da função somar:', somar(10, 5))
````

## 3. O que são Pacotes (*Packages*)?

Quando um projeto cresce muito e passa a ter dezenas de módulos `.py`, até mesmo os módulos ficam difíceis de organizar. Para resolver isso, criamos **Pacotes**.

- Um **Pacote** nada mais é do que uma **pasta (diretório)** que agrupa vários módulos separados por assunto (e pode conter subpacotes).
- Dentro de cada pasta de pacote, a convenção clássica do Python é incluir um arquivo especial chamado `__init__.py` (que pode ficar vazio ou expor as funções principais do pacote).

### Exemplo de estrutura de Pacote (como nos exercícios 111 e 112):
```text
meu_projeto/
 │
 ├── programa_principal.py
 └── utilidadescev/          <-- Pacote principal
      ├── __init__.py
      ├── moeda/             <-- Subpacote para cálculos monetários
      │    └── __init__.py
      └── dado/              <-- Subpacote para validação de dados
           └── __init__.py
```

Para importar a partir dessa estrutura no `programa_principal.py`:
```python
from utilidadescev import moeda, dado

preco = dado.leiaDinheiro('Digite o preço: R$ ')
moeda.resumo(preco, 20, 12)
```

## 4. Inspecionando Módulos (`dir`, `__doc__`, `__file__`)

Podemos inspecionar qualquer módulo ou objeto em Python para descobrir o que existe dentro dele:

````python
import math

# Listando os 10 primeiros atributos/funções públicos do módulo math:
publicos = [item for item in dir(math) if not item.startswith('_')]
print('Alguns recursos de math:', publicos[:10])

# Consultando a docstring de uma função do módulo:
print('Docstring de math.sqrt:', math.sqrt.__doc__)
````
