# Curso Python #08 - Utilizando Módulos

Nesta aula, aprendemos a expandir as funcionalidades nativas do Python através da **importação de módulos**.
- Um módulo é um arquivo contendo definições e instruções Python.
- Usamos o comando `import <modulo>` para importar tudo ou `from <modulo> import <funcao>` para importar membros específicos.


```python
# Importando o módulo de matemática padrão (math):
import math

num = 25
raiz = math.sqrt(num)
print(f'A raiz quadrada de {num} é {raiz:.2f}')
print(f'math.ceil(4.2) arredonda para cima: {math.ceil(4.2)}')
print(f'math.floor(4.8) arredonda para baixo: {math.floor(4.8)}')
print(f'Fatorial de 5 (math.factorial): {math.factorial(5)}')
```

```python
# Importando funções do módulo random:
from random import randint, choice, shuffle

sorteado = randint(1, 10)
print(f'Número aleatório entre 1 e 10: {sorteado}')

alunos = ['Ana', 'Bruno', 'Carlos', 'Daniela']
escolhido = choice(alunos)
print(f'Aluno sorteado para apagar o quadro: {escolhido}')

shuffle(alunos)
print(f'Ordem de apresentação embaralhada: {alunos}')
```

### 🌐 Cisco Python Essentials: Bytecode, sys.path e o Módulo `platform`

No estudo aprofundado de módulos (certificação PCAP/Cisco):
1. **Bytecode e `__pycache__`**: Quando um módulo `.py` é importado pela primeira vez, o interpretador compila o código em bytecode semi-compilado e o salva como `.pyc` dentro de `__pycache__` para acelerar execuções futuras.
2. **Caminho de Busca (`sys.path`)**: O Python procura módulos na seguinte ordem:
   - Diretório atual do script.
   - Variável de ambiente `PYTHONPATH`.
   - Bibliotecas padrão do Python.
   - Diretório de pacotes instalados (`site-packages`).
3. **Módulo `platform`**: Módulo padrão do Python para inspecionar as entranhas do sistema operacional e hardware da máquina:


```python
import platform
import sys

print('--- Informações do Sistema Operacional ---')
print(f'Sistema: {platform.system()}')
print(f'Versão detalhada: {platform.platform()}')
print(f'Arquitetura da máquina: {platform.machine()}')
print(f'Processador: {platform.processor()}')
print(f'Implementação do Python: {platform.python_implementation()} (versão {platform.python_version()})')

print('\n--- Primeiros 3 caminhos onde o Python busca módulos (sys.path) ---')
for p in sys.path[:3]:
    print(f'-> {p}')
```

### 🎲 Aprofundamento no `random`: Semente (`seed`) e Amostragem sem Reposição (`sample`)

- `random.seed(valor)`: O gerador de números aleatórios do Python é **pseudo-aleatório**. Definir uma semente faz com que a sequência de sorteios seja **100% reproduzível** (essencial para testes automatizados e simulações científicas).
- `random.choice(lista)`: Sorteia **1 elemento** de uma sequência.
- `random.sample(populacao, k)`: Sorteia **k elementos distintos sem repetição** (amostragem sem reposição).


```python
import random

# Reprodutibilidade com seed:
random.seed(42)
print('Sorteios com seed 42:', [random.randint(1, 100) for _ in range(3)])

# Amostragem sem reposição (loteria / megasena):
bilhete = random.sample(range(1, 61), 6)
print(f'6 dezenas sorteadas da Mega-Sena (sem repetição): {sorted(bilhete)}')
```
